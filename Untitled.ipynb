```python
import transformers
print(transformers.__version__)  # 4.44.2가 나와야 함

from genomic_benchmarks.loc2seq import download_dataset
path = download_dataset("human_nontata_promoters", version=0)
print(path)
```

```python
import os, random
random.seed(42)

def load_split(split, n_per_class=2000):
    seqs, labels = [], []
    split_dir = os.path.join(path, split)
    for cls in sorted(os.listdir(split_dir)):
        files = os.listdir(os.path.join(split_dir, cls))
        random.shuffle(files)
        for f in files[:n_per_class]:
            with open(os.path.join(split_dir, cls, f)) as fh:
                seqs.append(fh.read().strip().upper())
            labels.append(1 if cls == "positive" else 0)
    return seqs, labels

train_seqs, y_train = load_split("train", 2000)
test_seqs, y_test = load_split("test", 1000)
print(len(train_seqs), len(test_seqs), sum(y_train), sum(y_test))
```

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import roc_auc_score, average_precision_score

def kmers(seq, k=6):
    return " ".join(seq[i:i+k] for i in range(len(seq)-k+1))

vec = CountVectorizer()
X_train = vec.fit_transform([kmers(s) for s in train_seqs])
X_test = vec.transform([kmers(s) for s in test_seqs])

rf = RandomForestClassifier(n_estimators=300, n_jobs=-1, random_state=42)
rf.fit(X_train, y_train)
p_kmer = rf.predict_proba(X_test)[:, 1]

print("k-mer RF  AUROC:", round(roc_auc_score(y_test, p_kmer), 3))
print("k-mer RF  AUPRC:", round(average_precision_score(y_test, p_kmer), 3))
```

```python
import numpy as np

@torch.no_grad()
def embed_all(seqs, batch_size=32):
    out = []
    for i in range(0, len(seqs), batch_size):
        batch = tok(seqs[i:i+batch_size], return_tensors="pt", padding=True)
        ids = batch["input_ids"].cuda()
        mask = batch["attention_mask"].cuda()
        outs = model(ids, attention_mask=mask, encoder_attention_mask=mask, output_hidden_states=True)
        hidden = outs["hidden_states"][-1]
        m = mask.unsqueeze(-1).float()
        out.append(((hidden * m).sum(1) / m.sum(1)).cpu().numpy())
    return np.vstack(out)

E_train = embed_all(train_seqs)
E_test = embed_all(test_seqs)
print(E_train.shape, E_test.shape)
```

```python
rf2 = RandomForestClassifier(n_estimators=300, n_jobs=-1, random_state=42)
rf2.fit(E_train, y_train)
p_nt = rf2.predict_proba(E_test)[:, 1]

print("k-mer RF  AUROC:", round(roc_auc_score(y_test, p_kmer), 3), "| AUPRC:", round(average_precision_score(y_test, p_kmer), 3))
print("NT-v2 RF  AUROC:", round(roc_auc_score(y_test, p_nt), 3), "| AUPRC:", round(average_precision_score(y_test, p_nt), 3))
```

```python
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression

lr = make_pipeline(StandardScaler(), LogisticRegression(max_iter=2000))
lr.fit(E_train, y_train)
p_nt_lr = lr.predict_proba(E_test)[:, 1]
print("NT-v2 LR  AUROC:", round(roc_auc_score(y_test, p_nt_lr), 3), "| AUPRC:", round(average_precision_score(y_test, p_nt_lr), 3))
```

```python
@torch.no_grad()
def embed_layers(seqs, batch_size=32):
    layers = None
    for i in range(0, len(seqs), batch_size):
        batch = tok(seqs[i:i+batch_size], return_tensors="pt", padding=True)
        ids, mask = batch["input_ids"].cuda(), batch["attention_mask"].cuda()
        hs = model(ids, attention_mask=mask, encoder_attention_mask=mask, output_hidden_states=True)["hidden_states"]
        m = mask.unsqueeze(-1).float()
        pooled = [((h * m).sum(1) / m.sum(1)).cpu().numpy() for h in hs]
        layers = [[p] for p in pooled] if layers is None else [l + [p] for l, p in zip(layers, pooled)]
    return [np.vstack(l) for l in layers]

L_train, L_test = embed_layers(train_seqs), embed_layers(test_seqs)
layer_auc = []
for k in range(len(L_train)):
    clf = make_pipeline(StandardScaler(), LogisticRegression(max_iter=2000)).fit(L_train[k], y_train)
    auc = roc_auc_score(y_test, clf.predict_proba(L_test[k])[:, 1])
    layer_auc.append(auc)
    print(f"layer {k:2d}  AUROC: {auc:.3f}")
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import roc_curve

best = int(np.argmax(layer_auc))
clf_best = make_pipeline(StandardScaler(), LogisticRegression(max_iter=2000)).fit(L_train[best], y_train)
p_best = clf_best.predict_proba(L_test[best])[:, 1]

plt.figure(figsize=(6, 6))
for p, label in [(p_kmer, "6-mer + RF"), (p_nt, "NT-v2 last layer + RF"),
                 (p_nt_lr, "NT-v2 last layer + LR"), (p_best, f"NT-v2 layer {best} + LR")]:
    fpr, tpr, _ = roc_curve(y_test, p)
    plt.plot(fpr, tpr, label=f"{label} (AUROC={roc_auc_score(y_test, p):.3f})")
plt.plot([0, 1], [0, 1], "k--", lw=1)
plt.xlabel("False Positive Rate"); plt.ylabel("True Positive Rate")
plt.title("Human non-TATA promoter classification")
plt.legend(loc="lower right"); plt.tight_layout()
plt.savefig("roc_comparison.png", dpi=200); plt.show()
```
